```python
# Load and merge the IEEE-CIS transaction + identity data,
# then create the team's chronological train/validation/test split.

from pathlib import Path
from fraud.models.xgboost_baseline import load_merged, assign_split

RAW_DIR = Path("/content/drive/MyDrive/ieee-fraud-detection (1)")

df = load_merged(raw_dir=RAW_DIR)
split = assign_split(df)

print("Merged shape:", df.shape)
print("\nSplit source:", split.attrs.get("source"))
print(split.value_counts())

print("\nFraud rate:")
print(df["isFraud"].mean())
```

```python
# Separate the chronological training and validation data
# and define categorical vs numerical model features.

from fraud.categorical import CategoricalEncoder

train_df = df.loc[split == "train"].copy()
valid_df = df.loc[split == "valid"].copy()

encoder = CategoricalEncoder()

# These are the raw categorical features defined in Task #11.
categorical_raw_features = encoder.required_raw_cols

# These columns should not appear as ordinary numerical predictors.
excluded_columns = {
    "isFraud",
    "TransactionID",
    "TransactionDT",
    "day",
    "split",
    *categorical_raw_features,
}

# Keep all remaining genuinely numerical predictors.
numeric_features = [
    col
    for col in train_df.columns
    if col not in excluded_columns
    and pd.api.types.is_numeric_dtype(train_df[col])
]

print("Train rows:", len(train_df))
print("Validation rows:", len(valid_df))

print("\nRaw categorical features:", len(categorical_raw_features))
print(categorical_raw_features)

print("\nOriginal numerical features:", len(numeric_features))
print(numeric_features)
```

```python
# Fit the Task #11 categorical encoder on TRAINING DATA ONLY.
# Then apply the exact same learned mappings to validation data.

encoder.fit(train_df)

X_train_cat = encoder.transform(train_df)
X_valid_cat = encoder.transform(valid_df)

encoded_categorical_features = encoder.get_feature_names_out()

print("Encoded categorical features:", len(encoded_categorical_features))
print("Training categorical matrix:", X_train_cat.shape)
print("Validation categorical matrix:", X_valid_cat.shape)

print("\nFirst 20 encoded feature names:")
print(encoded_categorical_features[:20])
```

```python
# Convert the original numerical features to float32 and combine them
# with the Task #11 encoded categorical features.

import numpy as np
from scipy.sparse import csr_matrix, hstack

X_train_num = csr_matrix(
    train_df[numeric_features]
    .astype("float32")
    .to_numpy()
)

X_valid_num = csr_matrix(
    valid_df[numeric_features]
    .astype("float32")
    .to_numpy()
)

X_train = hstack(
    [X_train_num, X_train_cat],
    format="csr",
)

X_valid = hstack(
    [X_valid_num, X_valid_cat],
    format="csr",
)

feature_names = (
    numeric_features
    + list(encoded_categorical_features)
)

y_train = train_df["isFraud"].to_numpy()
y_valid = valid_df["isFraud"].to_numpy()

print("Final training matrix:", X_train.shape)
print("Final validation matrix:", X_valid.shape)
print("Total model features:", len(feature_names))
print("Training fraud rate:", y_train.mean())
print("Validation fraud rate:", y_valid.mean())
```

```python
# Train the Task #18 LightGBM baseline using class-imbalance weighting
# and early stopping on chronological validation AUPRC.

import time
import lightgbm as lgb

positive_count = y_train.sum()
negative_count = len(y_train) - positive_count
scale_pos_weight = negative_count / positive_count

print("Positive training cases:", int(positive_count))
print("Negative training cases:", int(negative_count))
print("scale_pos_weight:", round(scale_pos_weight, 3))

params = {
    "objective": "binary",
    "metric": "average_precision",
    "learning_rate": 0.05,
    "num_leaves": 256,
    "min_child_samples": 100,
    "feature_fraction": 0.5,
    "bagging_fraction": 0.8,
    "bagging_freq": 1,
    "lambda_l2": 1.0,
    "scale_pos_weight": scale_pos_weight,
    "verbosity": -1,
    "seed": 42,
    "feature_fraction_seed": 42,
    "bagging_seed": 42,
    "force_col_wise": True,
}

train_set = lgb.Dataset(
    X_train,
    label=y_train,
)

valid_set = lgb.Dataset(
    X_valid,
    label=y_valid,
    reference=train_set,
)

start = time.time()

model = lgb.train(
    params,
    train_set,
    num_boost_round=5000,
    valid_sets=[valid_set],
    valid_names=["valid"],
    callbacks=[
        lgb.early_stopping(200),
        lgb.log_evaluation(100),
    ],
)

training_seconds = time.time() - start

print("\nBest iteration:", model.best_iteration)
print("Training time:", round(training_seconds, 1), "seconds")
```

```python
# Generate validation fraud probabilities from the best LightGBM iteration.

valid_scores = model.predict(
    X_valid,
    num_iteration=model.best_iteration
)

print("Predictions:", len(valid_scores))
print("Min score:", valid_scores.min())
print("Max score:", valid_scores.max())
```

```python
# Evaluate LightGBM using Refilwe's Task #7 metric set.

import numpy as np
from sklearn.metrics import (
    average_precision_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    roc_curve,
)

def recall_at_review_rate(y_true, y_scores, review_rate):
    number_to_review = max(1, int(len(y_true) * review_rate))
    sorted_indices = np.argsort(y_scores)[::-1]
    top_transactions = sorted_indices[:number_to_review]

    fraud_found = y_true[top_transactions].sum()
    total_fraud = y_true.sum()

    return fraud_found / total_fraud


def evaluate_model(y_true, y_scores):
    y_pred = (y_scores >= 0.5).astype(int)

    auprc = average_precision_score(y_true, y_scores)
    precision = precision_score(y_true, y_pred, zero_division=0)
    recall = recall_score(y_true, y_pred, zero_division=0)
    f1 = f1_score(y_true, y_pred, zero_division=0)

    tn, fp, fn, tp = confusion_matrix(
        y_true,
        y_pred,
        labels=[0, 1]
    ).ravel()

    fpr = fp / (fp + tn)

    fpr_curve, tpr_curve, _ = roc_curve(y_true, y_scores)
    ks = np.max(np.abs(tpr_curve - fpr_curve))

    recall_1 = recall_at_review_rate(
        y_true,
        y_scores,
        0.01
    )

    recall_5 = recall_at_review_rate(
        y_true,
        y_scores,
        0.05
    )

    return {
        "AUPRC": auprc,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "FPR": fpr,
        "Recall@1%": recall_1,
        "Recall@5%": recall_5,
        "KS": ks,
    }


results = evaluate_model(y_valid, valid_scores)

for metric, value in results.items():
    print(f"{metric}: {value:.6f}")
```

```python
# Save Refilwe-style validation metrics for this LightGBM run.

evaluation_results = {
    metric: float(value)
    for metric, value in results.items()
}

with open(SAVE_DIR / "evaluation_metrics.json", "w") as f:
    json.dump(evaluation_results, f, indent=2)

print(json.dumps(evaluation_results, indent=2))
print("\nSaved to:", SAVE_DIR / "evaluation_metrics.json")
```

# Experiment 2 — LightGBM with Native Categorical Features

This experiment trains LightGBM using its native categorical handling instead of the explicit Task #11 one-hot/frequency encoding.

The same chronological train and validation rows are used so performance can be compared directly with Experiment 1.

```python
# Free the large encoded matrices from Experiment 1 before native LightGBM training.

import gc

del X_train
del X_valid
del X_train_num
del X_valid_num
del X_train_cat
del X_valid_cat
del train_set
del valid_set

gc.collect()

print("Experiment 1 matrices cleared from memory.")
```

```python
# Experiment 2: prepare raw numerical + native categorical features for LightGBM.
# Categorical levels are learned from the training split only.

import pandas as pd

NATIVE_EXCLUDE = {
    "isFraud",
    "TransactionID",
    "TransactionDT",
    "day",
    "split",
}

native_feature_names = [
    col
    for col in train_df.columns
    if col not in NATIVE_EXCLUDE
]

X_train_native = train_df[native_feature_names].copy()
X_valid_native = valid_df[native_feature_names].copy()

# Keep numerical values compact to reduce memory use.
for col in numeric_features:
    X_train_native[col] = X_train_native[col].astype("float32")
    X_valid_native[col] = X_valid_native[col].astype("float32")

# Treat all Task #11 categorical fields as genuine categorical variables.
# Categories are learned from TRAIN only.
for col in categorical_raw_features:
    train_values = train_df[col].astype("string")
    categories = pd.Index(train_values.dropna().unique())

    X_train_native[col] = pd.Categorical(
        train_values,
        categories=categories,
    )

    valid_values = valid_df[col].astype("string")

    # Categories not observed during training become missing in validation.
    valid_values = valid_values.where(
        valid_values.isin(categories)
    )

    X_valid_native[col] = pd.Categorical(
        valid_values,
        categories=categories,
    )

print("Native training matrix:", X_train_native.shape)
print("Native validation matrix:", X_valid_native.shape)

print("\nTotal raw model features:", len(native_feature_names))
print("Numerical features:", len(numeric_features))
print("Native categorical features:", len(categorical_raw_features))

print("\nExample categorical dtypes:")
print(
    X_train_native[
        ["card1", "card2", "addr1", "ProductCD", "DeviceInfo"]
    ].dtypes
)
```

```python
# Train Experiment 2: LightGBM with native categorical handling.

import time
import lightgbm as lgb

params_native = {
    "objective": "binary",
    "metric": "average_precision",
    "learning_rate": 0.05,
    "num_leaves": 256,
    "min_child_samples": 100,
    "feature_fraction": 0.5,
    "bagging_fraction": 0.8,
    "bagging_freq": 1,
    "lambda_l2": 1.0,
    "scale_pos_weight": scale_pos_weight,
    "verbosity": -1,
    "seed": 42,
    "feature_fraction_seed": 42,
    "bagging_seed": 42,
    "force_col_wise": True,
}

train_native_set = lgb.Dataset(
    X_train_native,
    label=y_train,
    categorical_feature=categorical_raw_features,
    free_raw_data=False,
)

valid_native_set = lgb.Dataset(
    X_valid_native,
    label=y_valid,
    categorical_feature=categorical_raw_features,
    reference=train_native_set,
    free_raw_data=False,
)

start_native = time.time()

native_model = lgb.train(
    params_native,
    train_native_set,
    num_boost_round=5000,
    valid_sets=[valid_native_set],
    valid_names=["valid"],
    callbacks=[
        lgb.early_stopping(200),
        lgb.log_evaluation(100),
    ],
)

native_training_seconds = time.time() - start_native

print("\nBest iteration:", native_model.best_iteration)
print(
    "Training time:",
    round(native_training_seconds, 1),
    "seconds"
)
```

```python
# Recreate the Task #7 evaluation functions after the runtime restart.

import numpy as np

from sklearn.metrics import (
    average_precision_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    roc_curve,
)

def recall_at_review_rate(y_true, y_scores, review_rate):
    number_to_review = max(1, int(len(y_true) * review_rate))

    sorted_indices = np.argsort(y_scores)[::-1]
    top_transactions = sorted_indices[:number_to_review]

    fraud_found = y_true[top_transactions].sum()
    total_fraud = y_true.sum()

    return fraud_found / total_fraud


def evaluate_model(y_true, y_scores):
    y_pred = (y_scores >= 0.5).astype(int)

    auprc = average_precision_score(y_true, y_scores)
    precision = precision_score(y_true, y_pred, zero_division=0)
    recall = recall_score(y_true, y_pred, zero_division=0)
    f1 = f1_score(y_true, y_pred, zero_division=0)

    tn, fp, fn, tp = confusion_matrix(
        y_true,
        y_pred,
        labels=[0, 1],
    ).ravel()

    fpr = fp / (fp + tn)

    fpr_curve, tpr_curve, _ = roc_curve(y_true, y_scores)
    ks = np.max(np.abs(tpr_curve - fpr_curve))

    recall_1 = recall_at_review_rate(
        y_true,
        y_scores,
        0.01,
    )

    recall_5 = recall_at_review_rate(
        y_true,
        y_scores,
        0.05,
    )

    return {
        "AUPRC": auprc,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "FPR": fpr,
        "Recall@1%": recall_1,
        "Recall@5%": recall_5,
        "KS": ks,
    }

print("Evaluator ready")
```
